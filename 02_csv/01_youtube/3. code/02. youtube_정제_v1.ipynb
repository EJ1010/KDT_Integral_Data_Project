```python
import re

def duration_to_seconds(value):
    if pd.isna(value):
        return pd.NA
    match = re.fullmatch(r"PT(?:(\d+)H)?(?:(\d+)M)?(?:(\d+)S)?", value)
    if not match:
        return pd.NA
    hours, minutes, seconds = (int(part) if part else 0 for part in match.groups())
    return hours * 3600 + minutes * 60 + seconds

df_relevance_audit["duration_sec"] = df_relevance_audit["duration"].map(duration_to_seconds)

df_relevance_audit["collected_at"] = pd.to_datetime(
    df_relevance_audit["collected_at"], utc=True
).dt.tz_convert("Asia/Seoul")
```

#### 3정규화(3NF)

`video_id` → `channel_id` → `channel` 이행 종속을 제거한다. 
<br> -> 영상 속성은 `video_id`에 직접 종속하므로 그대로 두고, 채널명만 분리한다.

- `youtube_analysis_channel_table(channel_id, channel)`
- `youtube_analysis_video_table(video_id, channel_id, publish_at, title, description, view_count, like_count, comment_count)`

원본 `df_analysis_merged`는 바꾸지 않는다.

```python
from pathlib import Path

channel_name_counts = df_analysis_merged.groupby("channel_id")["channel"].nunique()
if (channel_name_counts > 1).any():
    raise ValueError("channel_id마다 채널명이 하나씩이어야 합니다.")

youtube_analysis_channel_table = (
    df_analysis_merged[["channel_id", "channel"]]
    .drop_duplicates()
    .sort_values("channel_id", kind="stable")
    .reset_index(drop=True)
)

youtube_analysis_video_table = (
    df_analysis_merged[[
        "video_id",
        "channel_id",
        "publish_at",
        "title",
        "description",
        "view_count",
        "like_count",
        "comment_count",
    ]]
    .reset_index(drop=True)
)

output_dir = Path(r"C:/Users/user/DATA_MKT_260920/data.project/KDT_Integral_Data_Project/02_csv/01_youtube/영상/02. 정규화")
channel_path = output_dir / "youtube_analysis_channel_table.csv"
video_path = output_dir / "youtube_analysis_video_table.csv"

youtube_analysis_channel_table.to_csv(channel_path, index=False, encoding="utf-8-sig")
youtube_analysis_video_table.to_csv(video_path, index=False, encoding="utf-8-sig")

assert youtube_analysis_channel_table["channel_id"].is_unique
assert youtube_analysis_video_table["video_id"].is_unique
assert youtube_analysis_video_table["channel_id"].isin(
    youtube_analysis_channel_table["channel_id"]
).all()
assert len(youtube_analysis_video_table) == len(df_analysis_merged)

restored_channel = (
    youtube_analysis_video_table.merge(
        youtube_analysis_channel_table, on="channel_id", validate="many_to_one"
    )
    .set_index("video_id")
    .loc[df_analysis_merged["video_id"], "channel"]
    .reset_index(drop=True)
)
pd.testing.assert_series_equal(
    restored_channel,
    df_analysis_merged["channel"].reset_index(drop=True),
    check_names=False,
)

print(f"youtube_analysis_channel_table 저장: {channel_path.name} ({len(youtube_analysis_channel_table):,}행)")
print(f"youtube_analysis_video_table 저장: {video_path.name} ({len(youtube_analysis_video_table):,}행)")
```
